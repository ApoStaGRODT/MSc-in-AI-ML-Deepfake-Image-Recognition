```python
import os
import shutil
import random
from pathlib import Path
 
random.seed(42)
 
SOURCE_DIR = Path(r"C:\Users\user\Downloads\dataset")
DEST_DIR   = Path(r"C:\Users\user\DeepLearning\dataset")
 
SPLITS = {
    "train": {"real": 15000, "fake": 15000},
    "test":  {"real": 2500,  "fake": 2500},
    "valid": {"real": 2500,  "fake": 2500},
}
 
for split, classes in SPLITS.items():
    for cls, n in classes.items():
        src = SOURCE_DIR / split / cls
        dst = DEST_DIR / f"{split}-{cls}"
        dst.mkdir(parents=True, exist_ok=True)
 
        images = list(src.glob("*.jpg"))
        if len(images) < n:
            raise ValueError(f"Not enough images in {src}: found {len(images)}, need {n}")
 
        for img in random.sample(images, n):
            shutil.copy(img, dst / img.name)
 
print("Done.")
```
